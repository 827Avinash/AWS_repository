```python
# 2. find the largest digit of the number.
```

```python
num=int(input("Enter the number:"))
largest=0

while num>0:
    digit=num%10
    if digit>largest:
        largest=digit
    num=num//10
print("Largest digit in the given number is: ", largest)
```

```python
# 3. find the smallest digit of the number.
```

```python
num=int(input("Enter the number: "))
smallest=9

while num>0:
    digit=num%10
    if digit<smallest:
        smallest=digit
    num=num//10
print("Smallest number is: ", smallest)
    
```

```python
# 4. Find the reverse number given) by users.
```

```python
num=int(input("Enter the number: "))
reverse=0
while num>0:
    digit=num%10
    reverse=reverse*10+digit
    num=num//10
print("Reverse number of the given number is: ", reverse)
```

```python
# 5. Find the sum of the digits.
```

```python
num=int(input("Enter the number: "))
total=0
while num>0:
    digit=num%10
    total=total+digit
    num=num//10
print("Sum of the numbers are: ", total)
```

```python
# 6. Find a number is palindrom or not.
```

```python
num=int(input("Enter the number: "))
original=num
reverse=0

while num>0:
    digit=num%10
    reverse=reverse*10+digit
    num=num//10
if original == reverse:
    print("This is a palindrome")
else:
    print("This is not a palindrome")
```
